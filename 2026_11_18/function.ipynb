```python
# 1. 先在最左邊定義函數
def describe_pet(animal_type, pet_name):
    print(f"我有一隻 {animal_type}，名字叫 {pet_name}。")

# 2. 在「外面」（靠左對齊、沒有縮排）呼叫它
describe_pet('狗狗', '球球')
```

```python
def describe_pet(animal_type, pet_name):
    print(f"我養了一隻{animal_type}，牠叫做{pet_name}。")

# 【正確示範】關鍵字參數必須對應定義時的名稱 (pet_name)
describe_pet("狗", pet_name="旺財")
```

```python
# 定義函數時，幫 ice 和 sugar 設定預設值
def order_drink(drink, ice="正常冰", sugar="半糖"):
    print(f"點餐：{drink} ({ice} ／ {sugar})")

# 1. 只傳必要參數，其餘使用預設值
order_drink("紅茶")

# 2. 傳入第二個位置參數（改變冰塊，糖維持預設）
order_drink("紅茶", "去冰")

# 3. 使用引數名稱指定修改特定的項目（糖改無糖，冰維持預設）
order_drink("紅茶", sugar="無糖")

# 4. 透過引數名稱亂序傳入，同時修改冰塊與甜度
order_drink("綠茶", sugar="微糖", ice="少冰")
```

```python
def calc_bmi(weight, height, /):
    bmi = weight / (height ** 2)
    print(f"BMI 為：{bmi:.1f}")

# 正確呼叫方式（純用位置）
calc_bmi(80, 1.75)
```

```python
def send_email(to, *, subject, urgent=False):
    print(f"寄給：{to} | 主旨：{subject} | 急件：{urgent}")

# 正確呼叫方式（subject 與 urgent 必須寫出名稱）
send_email("student@example.com", subject="課堂通知", urgent=True)
```

```python
def register(id_no, /, name, *, role="student"):
    print(f"編號：{id_no} | 姓名：{name} | 身分：{role}")

# 呼叫範例（id_no 只能用位置，name 可用位置，role 必須用名稱）
register("A001", "王小明", role="teacher")
```

```python
def sum_scores(*scores):
    print("收到的資料：", scores, type(scores))
    return sum(scores)

# 測試不同的傳入數量
print(sum_scores(80, 90, 85))
print(sum_scores(80, 90))
print(sum_scores())
```

```python
def build_profile(name, **info):
    print("姓名：", name)
    print("其他資料：", info, type(info))

# 測試傳入額外的鍵值資料
build_profile("小華", city="台北", course="Python")
```
